# GETTER

```python
class momowallet:
    def __init__(self, owner_name, balance=0):
        self.owner_name = owner_name
        self._balance = balance

    @property
    def balance(self):
        return self._balance

account = momowallet("Ras", 1000)
print(f"{account.owner_name}'s wallet balance is {account.balance}")
```

# SETTER

```python
class momowallet:
    def __init__(self, owner_name, balance=0):
        self.owner_name = owner_name
        self._balance = balance

    @property
    def balance(self):
        return self._balance
    
    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = value

account = momowallet("Ras", 1000)
account.balance = 1000  
print(f"Welcome to Momo Wallet, {account.owner_name}! Your current balance is: {account.balance} ugx")
```
