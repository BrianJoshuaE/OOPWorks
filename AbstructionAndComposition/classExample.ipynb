```python
from abc import ABC, abstractmethod

class TransportMode(ABC):
    @abstractmethod
    def quote_fare(self, distance_km):
        pass

class BodaBoda(TransportMode):
    def quote_fare(self, distance_km):
        return distance_km * 1500

class Journey:
    def __init__(self, transport, distance_km):
        self.transport = transport  # composition
        self.distance_km = distance_km

    def cost(self):
        return self.transport.quote_fare(self.distance_km)

journey = Journey(BodaBoda(), 10)
print(journey.cost())
```
