```python
class FamilyMember:
    def __init__(self, eye_colour, home_village):
        self.eye_colour = eye_colour
        self.home_village = home_village

    def describe(self):
        print(f"my fam has {self.eye_colour} eyes and stays in {self.home_village}")

member = FamilyMember("red", "Osaka")
member.describe()
```
