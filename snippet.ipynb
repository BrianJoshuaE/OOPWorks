```python
class StudentGPA:
    def __init__(self, name, gpa):
        # Requirement: name should be public
        self.name = name
        
        # Initialize internal GPA with a default placeholder
        self._gpa = 0.0
        
        # Requirement: Constructor should validate the starting GPA
        # We use the setter to validate the initial value
        self.set_gpa(gpa)

    # Requirement: A getter should read the GPA
    def get_gpa(self):
        return self._gpa

    # Requirement: A setter should change the GPA only when valid
    def set_gpa(self, new_gpa):
        # Requirement: GPA should only be between 0.0 and 5.0
        if 0.0 <= new_gpa <= 5.0:
            self._gpa = new_gpa
        else:
            print(f"Error: Invalid GPA '{new_gpa}'. Must be between 0.0 and 5.0.")

# --- Testing the class as requested ---

print("--- Test 1: A valid GPA ---")
# Creating a student with a valid starting GPA
student1 = StudentGPA("John Doe", 3.5)
print(f"Student Name: {student1.name}")
print(f"Initial GPA: {student1.get_gpa()}")

print("\n--- Test 2: A valid update ---")
# Updating to another valid GPA
student1.set_gpa(4.2)
print(f"Updated GPA: {student1.get_gpa()}")

print("\n--- Test 3: An invalid update ---")
# Attempting to update to an invalid GPA (outside the 0.0 - 5.0 range)
student1.set_gpa(5.5)
# The GPA should remain unchanged (4.2)
print(f"Final GPA (after invalid update attempt): {student1.get_gpa()}")
```
