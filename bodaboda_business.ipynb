# Boda Boda Business

A simple business model for a boda boda owner.

```python
class bodaboda_business:
    """Manage the core details of a boda boda owner's business."""

    def __init__(self, fares, weight=0, season="normal"):
        self.fares = fares
        self.weight = weight
        self.season = season

    def set_fare(self, distance, amount):
        if amount < 0:
            raise ValueError("Fare cannot be negative")
        self.fares[distance] = amount

    def get_fare(self, distance):
        if distance not in self.fares:
            raise KeyError(f"No fare has been set for {distance}")
        return self.fares[distance]

    def update_weight(self, weight):
        if weight < 0:
            raise ValueError("Weight cannot be negative")
        self.weight = weight

    def change_season(self, season):
        if not season:
            raise ValueError("Season cannot be empty")
        self.season = season

    def calculate_fare(self, distance):
        fare = self.get_fare(distance)
        if self.season.lower() == "peak":
            return fare * 1.2
        return fare

    def business_summary(self):
        return {
            "fares": self.fares,
            "weight": self.weight,
            "season": self.season,
        }
```

```python
business = bodaboda_business(
    fares={"short": 100, "long": 250},
    weight=65,
    season="peak",
)

print("Short trip fare:", business.calculate_fare("short"))
business.update_weight(70)
business.change_season("normal")
print(business.business_summary())
```
