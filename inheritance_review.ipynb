```python
class Person:
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def introduce(self):
           print(f"Hello, my name is {self.name} and my email is {self.email}")

person = Person("John Doe", "john.doe@example.com")
person.introduce()

class Phone:
     def __init__(self, brand, airtime):
         self.brand = brand
         self.airtime = airtime
     def load_airtime(self, amount):
        self.airtime += amount
a = Phone("iPhone", 1000)
b = Phone("Tecno", 2000)

a.load_airtime(3000)
b.load_airtime(1000)
print(f"Current airtime for {a.brand}: {a.airtime}")
```

```python
class Person:
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def introduce(self):
           print(f"Hello, my name is {self.name} and my email is {self.email}")

person = Person("John Doe", "john.doe@example.com")
person.introduce()

class Phone:
     def __init__(self, brand, airtime):
         self.brand = brand
         self.airtime = airtime
     def load_airtime(self, amount):
        self.airtime=self.airtime + amount
a = Phone("iPhone", 1000)
b = Phone("Tecno", 2000)

a.load_airtime(3000)
b.load_airtime(1000)
print(f"Current airtime for {a.brand}: {a.airtime}")
print(f"Current airtime for {b.brand}: {b.airtime}")
```
