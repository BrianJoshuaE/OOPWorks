```python
class Course:
    def __init__(self, name):
        self._name = name

    def get_name(self):
        return self._name

    def set_name(self, name):
        if not name.strip(): # Check if empty or just whitespace
            print("Invalid course name.")
        else:
            self._name = name

course1 = Course("Object Oriented Programming")
print("Original:", course1.get_name())

# The code from the image
course1.set_name("Encapsulation")
print("Updated:", course1.get_name())

course1.set_name(" ")
print("Final:", course1.get_name())
```

```python
class StudentGPA:
    def __init__(self, name, gpa):
        # Requirement: name should be public
        self.name = name
        
        # Initialize internal GPA with a default placeholder
        self._gpa = 0.0
        
        # Requirement: Constructor should validate the starting GPA
        # We use the setter to validate the initial value
        self.set_gpa(gpa)

    # Requirement: A getter should read the GPA
    def get_gpa(self):
        return self._gpa

    # Requirement: A setter should change the GPA only when valid
    def set_gpa(self, new_gpa):
        # Requirement: GPA should only be between 0.0 and 5.0
        if 0.0 <= new_gpa <= 5.0:
            self._gpa = new_gpa
        else:
            print(f"Error: Invalid GPA '{new_gpa}'. Must be between 0.0 and 5.0.")

# --- Testing the class as requested ---

print("--- Test 1: A valid GPA ---")
# Creating a student with a valid starting GPA
student1 = StudentGPA("John Doe", 3.5)
print(f"Student Name: {student1.name}")
print(f"Initial GPA: {student1.get_gpa()}")

print("\n--- Test 2: A valid update ---")
# Updating to another valid GPA
student1.set_gpa(4.2)
print(f"Updated GPA: {student1.get_gpa()}")

print("\n--- Test 3: An invalid update ---")
# Attempting to update to an invalid GPA (outside the 0.0 - 5.0 range)
student1.set_gpa(5.5)
# The GPA should remain unchanged (4.2)
print(f"Final GPA (after invalid update attempt): {student1.get_gpa()}")
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        # Public attribute
        self.owner = owner
        # Internal attribute (cannot be changed directly from outside)
        self._balance = balance

    # Behavior to add money
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Deposited {amount}. New balance: {self._balance}")
        else:
            print("Error: Deposit must be positive.")

    # Behavior to take out money
    def withdraw(self, amount):
        if amount <= 0:
            print("Error: Withdrawal must be positive.")
        elif amount > self._balance:
            print("Error: Cannot withdraw more than available balance.")
        else:
            self._balance -= amount
            print(f"Withdrew {amount}. New balance: {self._balance}")

    # Getter to safely read the balance
    def get_balance(self):
        return self._balance


# --- Demonstration of valid and invalid operations ---
print("--- Valid Operations ---")
account = BankAccount("Alice", 100)
account.deposit(50)       # Valid deposit
account.withdraw(30)      # Valid withdrawal

print("\n--- Invalid Operations ---")
account.deposit(-20)      # Invalid: negative
account.withdraw(-10)     # Invalid: negative
account.withdraw(500)     # Invalid: exceeds balance

print(f"\nFinal Balance: {account.get_balance()}")
```
