```python
class BodaBoda:
    def transport(self, passenger):
        return f"{passenger} rides pillion to Mukono Town."
 
class Matatu:
    def transport(self, passenger):
        return f"{passenger} squeezes in for the Kampala route."
 
for vehicle in [BodaBoda(), Matatu()]:
    print(vehicle.transport("Auma Grace"))
```

```python
class PhoneAlarm:
    def __init__(self, snooze_count):
        self.snooze_count = snooze_count

    def wake_me_up(self):
        return "screen at full volume"

print(PhoneAlarm(3).wake_me_up())
```
