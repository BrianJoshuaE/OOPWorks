```python
class BodaBoda:
    def transport(self, passenger):
        return f"{passenger} rides pillion to Mukono Town."
 
class Matatu:
    def transport(self, passenger):
        return f"{passenger} squeezes in for the Kampala route."
 
for vehicle in [BodaBoda(), Matatu()]:
    print(vehicle.transport("Auma Grace"))
```

```python
class PhoneAlarm:
    def __init__(self, snooze_count):
        self.snooze_count = snooze_count

    def wake_me_up(self):
        return "screen at full volume"

class RoomMate:
    def __init__(self, name):
        self.name = name

    def wake_me_up(self):
        return f"{self.name} slaps you from sleep"

class MotherCall:
    def __init__(self, name):
        self.name = name

    def wake_me_up(self):
        return f"{self.name} calls you to wake up"

class SunlightThroughTheWindow:
    def wake_me_up(self):
        return "sunlight through the window wakes you up"

morning = [
    PhoneAlarm(5),
    RoomMate("Epaja"),
    MotherCall("Peace"),
    SunlightThroughTheWindow()
]

for wake_up_method in morning:
    print(wake_up_method.wake_me_up())

```
